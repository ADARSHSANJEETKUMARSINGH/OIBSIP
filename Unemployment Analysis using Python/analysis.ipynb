## 1. Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Reading the data

```python
df1 = pd.read_csv("data/Unemployment in India.csv")
df2 = pd.read_csv("data/Unemployment_Rate_upto_11_2020.csv")
```

## 3.Data cleaning

```python
df1.columns = df1.columns.str.strip()
```

```python
df1=df1.dropna()
```

```python
df1["Estimated Unemployment Rate (%)"]
```

## 4. Exploratory Data Analysis

```python
plt.figure(figsize=(15,10))
sns.countplot(y="Region", data=df1, hue="Region", dodge=False)
plt.title("Number of Records per State")
plt.show()
```

```python
avg_unemployment_rate = df1.groupby('Region')['Estimated Unemployment Rate (%)'].mean()

state_with_highest_unemployment = avg_unemployment_rate.idxmax()
high_unemployment_rate = avg_unemployment_rate.max()

state_with_lowest_unemployment = avg_unemployment_rate.idxmin()
low_unemployment_rate = avg_unemployment_rate.min()

print(f"state with high unemployment : {state_with_highest_unemployment}")
print(f"high unemployment rate : {high_unemployment_rate}")

print(f"state with low unemployment : {state_with_lowest_unemployment}")
print(f"low unemployment rate : {low_unemployment_rate}")
```

```python
regions = avg_unemployment_rate.index
avg_rates = avg_unemployment_rate.values

plt.figure(figsize=(10, 6))
plt.bar(regions, avg_rates, color='skyblue')
plt.xlabel('Region')
plt.ylabel('Average Unemployment Rate')
plt.title('Average Unemployment Rate by Region')
plt.xticks(rotation=90)  
plt.tight_layout()  
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x='Region', y='Estimated Unemployment Rate (%)', data=df1, palette="Set1")
plt.title("Unemployment Rate Distribution by Region")
plt.xlabel("State")
plt.ylabel("Estimated Unemployment Rate")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x='Region', y='Estimated Labour Participation Rate (%)', data=df1, palette="hsv")
plt.title("Labour Participation Rate by Region")
plt.xlabel("Region")
plt.ylabel("Labour Participation Rate")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x='Region', y='Estimated Employed', data=df1, palette="brg")
plt.title("Estimated Employed by Region")
plt.xlabel("Region")
plt.ylabel("Estimated Employed")
plt.xticks(rotation=90)
plt.show()
```

```python
sns.histplot(data=df1, x="Estimated Unemployment Rate (%)", kde=True, color="navy")
plt.show()
```

```python
sns.histplot(data=df1, x="Estimated Employed", kde=True, color="brown")
plt.show()
```

```python
sns.histplot(data=df1, x="Estimated Labour Participation Rate (%)", kde=True, color="darkorange")
plt.show()
```
